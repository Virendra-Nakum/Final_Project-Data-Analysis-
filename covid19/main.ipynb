```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

print("========== COVID-19 DATA ANALYSIS ==========")
try : 
    
    path = input("Enter Dataset Path (e.g. data/sales_data.csv): ").strip()
    data = pd.read_csv(path) 
    print("Data Loaded Sucesfully !")
    
except :
    print('File Not Found')
```

```python

print("========== EXPLORE DATASET ==========")
print("1. Display First 5 Rows")
print("2. Display Last 5 Rows")
print("3. Display Column Names")
print("4. Display Data Types")
print("5. Display Dataset Information")
print("6. Display Statistical Summary")

try :
    choice = input("Enter Your Choice :- ")

    if choice == '1' :
        print(data.head())

    elif choice == '2' :
        print(data.tail())

    elif choice == '3' :
        print(data.columns)

    elif choice == '4' :
        print(data.dtypes)

    elif choice == '5' :
        print(data.info())

    elif choice == '6' :
        print(data.describe())

    else :
        print("Enter Valid Choice !")

except :
    print("Load Dataset First")
```

```python
print("========== CLEAN DATASET ==========")
print("1. Handle Missing Values")
print("2. Remove Duplicate Records")
print("3. Convert Data Types")
print("4. Clean Column Names")
try:
    choice = int(input("Enter Your Choice :-"))

    if choice == 1 :
        print(data.isnull().sum())
        print(data.ffill())

    elif choice == 2 :
        print(data.drop_duplicates())

    elif choice == 3 :
        data['Date'] = pd.to_datetime(data['Date'])
        print(data.dtypes)

    elif choice == 4 :
        print(" OLD COLUMNS ")
        print(data.columns)
        colm = input("Enter Column Name :- ")

        if colm in data.columns:
            new = input("Enter New Column Name :- ")
            data.rename(columns={colm: new}, inplace=True)
            print("Column Name Updated Successfully!")
        else:
            print("Column Not Found!")

        print(data.columns)

    else :
        print("Enter Valid Choice !")
except :
    print('Load Dataset First !')
```

```python
print("========== DATA ANALYSIS ==========")
print("1. Total Confirmed Cases")
print("2. Total Deaths")
print("3. Total Recovered Cases")
print("4. Active Cases")
print("5. Country-wise Analysis")
print("6. Highest and Lowest Cases")
print("7. Death Rate and Recovery Rate")

try :
    choice = input("Enter Choice  :- ") 
    if choice == "1" :
        print(f"Total Confirmed Cases is : {data['Confirmed'].sum()}")

    elif choice == "2" :
        print(f"Total Deaths Cases is : {data['Deaths'].sum()}")

    elif choice == '3' :
        print(f"Total Recovered Cases is : {data['Recovered'].sum()}")

    elif choice == '4':
        print(f"Total Active Cases: {data['Active'].sum()}")

    elif choice == '5':
        print(data.groupby('Country_Region')[['Confirmed', 'Deaths', 'Recovered', 'Active']].sum())

    elif choice == '6':
        print("Minimum Values by Country:")
        print(data.groupby('Country_Region')[['Confirmed', 'Deaths', 'Recovered', 'Active']].min())

        print("Maximum Values by Country:")
        print(data.groupby('Country_Region')[['Confirmed', 'Deaths', 'Recovered', 'Active']].max())

    elif choice == '7':
        data["Death Rate (%)"] = ((data["Deaths"] / data["Confirmed"]) * 100).round(2)
        data["Recovery Rate (%)"] = ((data["Recovered"] / data["Confirmed"]) * 100).round(2)

        print(data[["Country_Region", "Death Rate (%)", "Recovery Rate (%)"]])

    else:
        print("Invalid Choice!")
        
except :
    print("Load Dataset first !")
```

```python
print("========== DATA VISUALIZATION ==========")
print("1. Bar Plot")
print("2. Line Plot")
print("3. Scatter Plot")
print("4. Histogram")
print("5. Pie Chart")
print("6. Heatmap")
try :
    choice = input("Enter Your Choice :- ")
    sns.set_theme(style="whitegrid", palette="Set2")

    if choice == '1' :

        grouped = data.groupby("Country_Region")["Deaths"].sum().reset_index().dropna()
        grouped = grouped[grouped["Deaths"] > 0]
        grouped = grouped.sort_values("Deaths")
        plt.figure(figsize=(14, 20))
        sns.barplot(
            data=grouped,
            x="Deaths",
            y="Country_Region",
            color="skyblue",
            edgecolor="black",
            linewidth=0.3)
        
        plt.title("Deaths by All Countries")
        plt.xlabel("Deaths")
        plt.ylabel("Country_Region")
        plt.tick_params(axis="y", labelsize=7)
        plt.tight_layout()
        plt.show()

    elif choice == '2' :

        temp = data[["Date", "Confirmed"]].copy()
        temp["Date"] = pd.to_datetime(temp["Date"])
        temp = temp.dropna()
        temp = temp[temp["Confirmed"] > 0]
        grouped = temp.groupby("Date")["Confirmed"].sum().reset_index()

        plt.figure(figsize=(10, 5))
        sns.lineplot(
            data=grouped,
            x="Date",
            y="Confirmed",
            color="royalblue",
            marker="o",
            errorbar=None)

        plt.title("Confirmed Cases Over Time")
        plt.xlabel("Date")
        plt.ylabel("Confirmed Cases")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

    elif choice == '3' :
        temp = data[["Confirmed", "Deaths"]].dropna()
        temp = temp[(temp["Confirmed"] > 0) & (temp["Deaths"] > 0)]
        plt.figure(figsize=(10, 5))
        sns.scatterplot(
            data=temp,
            x="Confirmed",
            y="Deaths",
            hue="Deaths",
            palette="viridis",
            alpha=0.7,
            s=60)
        
        plt.title("Confirmed Cases vs Deaths")
        plt.xlabel("Confirmed Cases")
        plt.ylabel("Deaths")
        plt.tight_layout()
        plt.show()

    elif choice == '4' :
        temp = data[["Recovered"]].dropna()
        temp = temp[temp["Recovered"] > 0]
        plt.figure(figsize=(10, 5))
        sns.histplot(
            data=temp,
            x="Recovered",
            bins=20,
            kde=True,
            color="orange",
            edgecolor="black",
            alpha=0.7)
        
        plt.title("Recovered Cases Distribution")
        plt.xlabel("Recovered Cases")
        plt.ylabel("Frequency")
        plt.tight_layout()
        plt.show()

    elif choice == '5'  :
            
        values = data.groupby("Country_Region")["Deaths"].sum()
        values = values.dropna()
        values = values[values > 0]
        values = values.sort_values(ascending=False)
        plt.figure(figsize=(12, 9))
        plt.pie(
            values,
            labels=values.index,
            autopct="%1.1f%%",
            startangle=90,
            colors=sns.color_palette("pastel", len(values)),
            wedgeprops={"edgecolor": "white"})
        
        plt.title("Deaths Percentage by Country")
        plt.axis("equal")
        plt.tight_layout()
        plt.show()

    elif choice == '6' :
        numeric_data = data.select_dtypes(include="number")
        plt.figure(figsize=(10, 5))
        sns.heatmap(
            numeric_data.corr(),
            annot=True,
            fmt=".2f",
            cmap="coolwarm",
            linewidths=0.5,
            vmin=-1,
            vmax=1)
        
        plt.title("COVID-19 Correlation Heatmap")
        plt.tight_layout()
        plt.show()

    else:
        print("Something Went Wrong !")

except :
    print("Load Dataset First !")
```

```python
print("========== ANALYSIS RESULTS ==========")
print("1. Display Overall Summary")
print("2. Display COVID-19 Performance")
print("3. Display Country Analysis")
print("4. Display Trend Analysis")
print("5. Display Final Report")

try:
    choice = input("Enter Your Choice: ")

    if choice == "1":
        print("========== OVERALL SUMMARY ==========")
        print("Total Records:", len(data))
        print("Total Confirmed Cases:", data["Confirmed"].sum())
        print("Total Deaths:", data["Deaths"].sum())
        print("Total Recovered:", data["Recovered"].sum())
        print("Total Active Cases:", data["Active"].sum())

    elif choice == "2":
        print("========== COVID-19 PERFORMANCE ==========")
        highest_cases_country = data.loc[data["Confirmed"].idxmax(), "Country_Region"]
        lowest_cases_country = data.loc[data["Confirmed"].idxmin(), "Country_Region"]

        print("Country with Highest Confirmed Cases:", highest_cases_country)
        print("Highest Confirmed Cases:", data["Confirmed"].max())
        print("Country with Lowest Confirmed Cases:", lowest_cases_country)
        print("Lowest Confirmed Cases:", data["Confirmed"].min())

    elif choice == "3":
        print("========== COUNTRY ANALYSIS ==========")
        country_cases = (data.groupby("Country_Region")["Confirmed"].sum().sort_values(ascending=False))
        print(country_cases.head(10))

        print("Country with Highest Confirmed Cases:", country_cases.idxmax())
        print("Country with Lowest Confirmed Cases:", country_cases.idxmin())

    elif choice == "4":
        print("========== TREND ANALYSIS ==========")
        daily_cases = data.groupby("Date")["Confirmed"].sum()
        print(daily_cases)

        print("Date with Highest Confirmed Cases:", daily_cases.idxmax())
        print("Date with Lowest Confirmed Cases:", daily_cases.idxmin())

    elif choice == "5":
        print("========== FINAL COVID-19 REPORT ==========")
        print(data[["Confirmed", "Deaths", "Recovered", "Active"]].describe())

        print("Overall Analysis:")

        if data["Deaths"].sum() > 0:
            print("COVID-19 deaths were recorded during this period.")
        else:
            print("No COVID-19 deaths were recorded during this period.")

        if data["Recovered"].sum() > 0:
            print("COVID-19 recoveries were recorded during this period.")
        
        else:
            print("No COVID-19 recoveries were recorded during this period.")

    else:
        print("Invalid Choice!")

except :
    print("Load Dataset First")
```
