```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("========== GLOBAL HAPPINESS ANALYSIS ==========")

try:
    load = input("Enter Dataset Path (e.g data/sales.csv): ").strip()
    data = pd.read_csv(load,sep=';')  
    print('Dataset Loaded Successfully ')
    
except :
    print("File Not Found !")
```

```python

print("========== EXPLORE DATASET ==========")
print("1. Display First 5 Rows")
print("2. Display Last 5 Rows")
print("3. Display Column Names")
print("4. Display Data Types")
print("5. Display Dataset Information")
print("6. Display Statistical Summary")

try :
    choice = input("Enter Your Choice :- ")

    if choice == '1' :
        print(data.head())

    elif choice == '2' :
        print(data.tail())

    elif choice == '3' :
        print(data.columns)

    elif choice == '4' :
        print(data.dtypes)

    elif choice == '5' :
        print(data.info())

    elif choice == '6' :
        print(data.describe())

    else :
        print("Enter Valid Choice !")

except :
    print("Load Dataset First")
```

```python
print("========== CLEAN DATASET ==========")
print("1. Check Missing Values")
print("2. Handle Missing Values")
print("3. Remove Duplicate Rows")  

try :
    choice = input("Enter Your Choice :- ")

    if choice == '1' :
        print(data.isnull().sum())

    elif choice == '2' :
        print(data.ffill())

    elif choice == '3' :
        print(data.drop_duplicates())

    else :
        print("Invalid Choice !")

except : 
    print("Load Dataset First")
 
```

```python
print("========== GLOBAL HAPPINESS DATA ANALYSIS ==========")
print("1. Calculate Average Happiness Score")
print("2. Find Happiest and Least Happy Countries")
print("3. Compare Average Happiness by Region")
print("4. Analyze GDP per Capita and Happiness")
print("5. Analyze Social Support and Happiness")
print("6. Analyze Healthy Life Expectancy and Happiness")
print("7. Analyze Happiness Trends by Year")
 
columns = ["Happiness score","GDP per capita","Social support", "Healthy life expectancy","Freedom to make life choices","Generosity","Perceptions of corruption"]

for col in columns:
    data[col] = (data[col].astype(str).str.replace(",", ".", regex=False).str.strip().astype(float))
 
data["Ranking"] = pd.to_numeric(data["Ranking"]).astype("Int64")
data["Year"] = pd.to_numeric(data["Year"]).astype("Int64")

try :
    choice = input("Enter Your Choice: ")

    if choice == '1':
        print(data["Happiness score"].mean())

    elif choice == '2':
        row = data["Happiness score"].idxmin()
        row1 = data["Happiness score"].idxmax()

        print("MINIMUM:", data.loc[row, ["Country", "Happiness score"]])
        print("MAXIMUM:", data.loc[row1, ["Country", "Happiness score"]])

    elif choice == '3':
        print(data.groupby("Regional indicator")["Happiness score"].mean())

    elif choice == '4':
        result = data["GDP per capita"].corr(data["Happiness score"])  # corr = correstion mtb -1 to 1 
        print("GDP and Happiness Correlation:", result)

    elif choice == '5':
        result = data["Social support"].corr(data["Happiness score"])
        print("Social Support and Happiness Correlation:", result)

    elif choice == '6':
        result = data["Healthy life expectancy"].corr(data["Happiness score"])
        print("Life Expectancy and Happiness Correlation:", result)

    elif choice == '7':
        yearly = data.groupby("Year")["Happiness score"].mean()
        print("Year-wise Average Happiness Score:", yearly) 
        print("Change from Previous Available Year:", yearly.diff())
      
    else:
        print("Enter A Valid Choice!")

except :    
    print("Load Dataset First")
```

```python
print("========== GLOBAL HAPPINESS DATA VISUALIZATION ==========")
print("1. Year-wise Happiness Line Chart")
print("2. Average Happiness by Region Bar Chart")
print("3. GDP per Capita vs Happiness Scatter Plot")
print("4. Happiness Score Histogram")
print("5. Social Support vs Happiness Scatter Plot")
print("6. Correlation Heatmap")

try:
    choice = input("Enter Your Choice :- ").strip()

    if choice == '1':
        data1 = data.groupby("Year")["Happiness score"].mean().reset_index()

        plt.figure(figsize=(10, 5))

        sns.lineplot(
            data=data1,
            x="Year",
            y="Happiness score",
            marker="o",
            hue = True)

        plt.xticks(data1["Year"], rotation=45)
        plt.title("Year-wise Average Happiness Score")
        plt.xlabel("Year")
        plt.ylabel("Average Happiness Score")
        plt.tight_layout()
        plt.show()

    elif choice == '2' : 
        data2 = (
            data.groupby("Regional indicator")["Happiness score"].mean().reset_index())

        plt.figure(figsize=(10, 20))
        sns.barplot(
            data=data2,
            x="Regional indicator",
            y="Happiness score",
            hue="Regional indicator",
            palette="viridis",
            legend= True)

        plt.title("Average Happiness by Region")
        plt.xlabel("Region")
        plt.ylabel("Average Happiness Score")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()

    elif choice == '3' :
        data3 = data.groupby('GDP per capita')['Happiness score'].sum().reset_index()

        plt.figure(figsize=(10,5))
        sns.scatterplot(
            data= data3,
            x= 'GDP per capita',
            y= 'Happiness score' ,
            hue = 'GDP per capita',
            legend=  True)
        plt.title('GDP per Capita vs Happiness Scatter')
        plt.show()
    
    elif choice == '4' : 
        plt.figure(figsize=(10, 5))
        sns.histplot(data=data,x="Happiness score",bins=20)
        plt.title("Distribution of Happiness Scores")
        plt.xlabel("Happiness Score")
        plt.ylabel("Number of Countries")
        plt.tight_layout()
        plt.show()

    elif choice == '5' :
        data5 = data.groupby('Social support')['Happiness score'].sum().reset_index()

        plt.figure(figsize=(10,5))
        sns.scatterplot(
            data= data5,
            x= 'Social support',
            y= 'Happiness score' ,
            hue = 'Social support',
            legend=  True)
        
        plt.title('GDP per Capita vs Happiness Scatter')
        plt.show()

    elif choice == '6' : 
        numeric_data = data.select_dtypes(include="number").corr()


        plt.figure(figsize=(12, 8))

        sns.heatmap( 
            numeric_data,
            annot=True,
            fmt=".2f",
            cmap="coolwarm", 
            linewidths=0.5, 
        )

        plt.title("Correlation Heatmap")
        plt.tight_layout()
        plt.show()

    else :
        print("Please Enter Valid Choice !")

except :
    print('Load Dataset First !')
```

```python
print("========== ANALYSIS RESULTS ==========")
print("1. Display Overall Summary")
print("2. Display Happiness Performance")
print("3. Display Regional Analysis")
print("4. Display Factor Analysis")
print("5. Display Final Report")
try:
    choice = input("Enter Your Choice: ")

    if choice == "1":
        print("========== OVERALL SUMMARY ==========")
        print("Total Countries:", len(data))
        print("Average Happiness Score:", data["Happiness score"].mean())
        print("Highest Happiness Score:", data["Happiness score"].max())
        print("Lowest Happiness Score:", data["Happiness score"].min())

    elif choice == "2":
        print("========== HAPPINESS PERFORMANCE ==========")
        happiest_country = data.loc[data["Happiness score"].idxmax(), "Country"]
        least_happy_country = data.loc[data["Happiness score"].idxmin(), "Country"]

        print("Happiest Country:", happiest_country)
        print("Happiness Score:", data["Happiness score"].max())
        print("Least Happy Country:", least_happy_country)
        print("Happiness Score:", data["Happiness score"].min())

    elif choice == "3":
        print("========== REGIONAL ANALYSIS ==========")
        regional_average = data.groupby("Regional indicator")["Happiness score"].mean()

        print(regional_average)

        print("Highest Average Happiness Region:", regional_average.idxmax())
        print("Lowest Average Happiness Region:", regional_average.idxmin())

    elif choice == "4":
        print("========== FACTOR ANALYSIS ==========")
        print("GDP per Capita Correlation:", data["GDP per capita"].corr(data["Happiness score"]))
        print("Social Support Correlation:", data["Social support"].corr(data["Happiness score"]))
        print("Healthy Life Expectancy Correlation:", data["Healthy life expectancy"].corr(data["Happiness score"]))

    elif choice == "5":
        print("========== FINAL HAPPINESS REPORT ==========")
        print(data[["Happiness score", "GDP per capita", "Social support", "Healthy life expectancy"]].describe())

        print("Overall Analysis:")

        if data["GDP per capita"].corr(data["Happiness score"]) > 0:
            print("GDP per capita aur Happiness score ke beech positive relationship hai.")
        else:
            print("GDP per capita aur Happiness score ke beech negative relationship hai.")

        if data["Social support"].corr(data["Happiness score"]) > 0:
            print("Social support aur Happiness score ke beech positive relationship hai.")
        else:
            print("Social support aur Happiness score ke beech negative relationship hai.")

        if data["Healthy life expectancy"].corr(data["Happiness score"]) > 0:
            print("Healthy life expectancy aur Happiness score ke beech positive relationship hai.")
        else:
            print("Healthy life expectancy aur Happiness score ke beech negative relationship hai.")

    else:
        print("Invalid Choice!")

except :
    print("Load Dataset First !")
```
