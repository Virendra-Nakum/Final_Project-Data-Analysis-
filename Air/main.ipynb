```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

print("========== COVID-19 DATA ANALYSIS ==========")
try : 
    
    path = input("Enter Dataset Path (e.g. data/sales_data.csv): ").strip()
    data = pd.read_csv(path,sep=';') 
    print("Data Loaded Sucesfully !")
    
except :
    print('File Not Found')
```

```python

print("========== EXPLORE DATASET ==========")
print("1. Display First 5 Rows")
print("2. Display Last 5 Rows")
print("3. Display Column Names")
print("4. Display Data Types")
print("5. Display Dataset Information")
print("6. Display Statistical Summary")

try :
    choice = input("Enter Your Choice :- ")

    if choice == '1' :
        print(data.head())

    elif choice == '2' :
        print(data.tail())

    elif choice == '3' :
        print(data.columns)

    elif choice == '4' :
        print(data.dtypes)

    elif choice == '5' :
        print(data.info())

    elif choice == '6' :
        print(data.describe())

    else :
        print("Enter Valid Choice !")

except :
    print("Load Dataset First")
```

```python
print("========== CLEAN DATASET ==========")
print("1. Check Missing Values")
print("2. Handle Missing Values")
print("3. Remove Duplicate Rows")  

try :
    choice = input("Enter Your Choice :- ")

    if choice == '1' :
        print(data.isnull().sum())

    elif choice == '2' :
        print(data.ffill())

    elif choice == '3' :
        print(data.drop_duplicates())

    else :
        print("Invalid Choice !")

except : 
    print("Load Dataset First")
```

```python
print("========== DATA ANALYSIS ==========")
print("1. Total Countries")
print("2. Total Cities")
print("3. Total Pollutants")
print("4. Country-wise Analysis")
print("5. City-wise Analysis")
print("6. Highest and Lowest Pollution Value")
print("7. Average Pollution Value by Country")
print("8. Most Common Pollutant")
print("9. Latest Updated Records")


try:
    choice = input("Enter Choice :- ")

    if choice == '1':
        print("Total Countries Is :", data['Country'].nunique())

    elif choice == '2':
        print("Total Cities Is :", data['City'].nunique())

    elif choice == '3':
        print("Total Pollutants Is :", data['Pollutant'].nunique())

    elif choice == '4':
        print(data.groupby('Country')['Value'].agg(['count', 'mean', 'max', 'min']))

    elif choice == '5':
        print(data.groupby('City')['Value'].agg(['count', 'mean', 'max', 'min']))

    elif choice == '6':
        print("Highest Pollution Value :")
        print(data['Value'].max())

        print("Lowest Pollution Value :")
        print(data['Value'].min())

    elif choice == '7':
        print(data.groupby('Country')['Value'].mean())

    elif choice == '8':
        print(data['Pollutant'].value_counts())

    elif choice == '9':
        print("Latest Updated Record :")
        print(data['Last Updated'].max())

    else :
        print("Invalid Choice !")

except :
    print("Load Dataset First !")
```

```python
print("========== DATA VISUALIZATION ==========")
print("1. Bar Plot")
print("2. Line Plot")
print("3. Scatter Plot")
print("4. Histogram")
print("5. Pie Chart")
print("6. Heatmap")
print("7. Subplot")


try:
    choice = input("Enter Your Choice :- ")

    if choice == '1':
        data1 = data.groupby('Country')['Value'].mean().reset_index()
        plt.figure(figsize=(20, 10))
        sns.barplot(
            data=data1,
            x='Country',
            y='Value',
            hue='Country',
            legend=False
        )

        plt.title('Country Wise Average Pollution Value')
        plt.xlabel('Country')
        plt.ylabel('Average Pollution Value')
        plt.xticks(rotation=90)
        plt.tight_layout()
        plt.show()

    elif choice == '2':

        data['Last_Updated'] = pd.to_datetime(data['Last_Updated'], errors='coerce')
        data2 = data.groupby('Last_Updated')['Value'].mean().reset_index()
        plt.figure(figsize=(15, 6))

        sns.lineplot(
            data=data2,
            x='Last_Updated',
            y='Value',
            color='green')
        plt.title('Pollution Value Over Time')
        plt.xlabel('Last_Updated')
        plt.ylabel('Average Pollution Value')
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

    elif choice == '3':

        plt.figure(figsize=(15, 6))
        sns.scatterplot(
            x=data.index,
            y='Value',
            data=data,
            alpha=0.3,
            color='purple')
        plt.title('Pollution Value Scatter Plot')
        plt.xlabel('Record Index')
        plt.ylabel('Pollution Value')
        plt.tight_layout()
        plt.show()

    elif choice == '4':

        plt.figure(figsize=(12, 6))
        sns.histplot(data['Value'], bins=30, kde=True, color='skyblue')
        plt.title('Pollution Value Distribution')
        plt.xlabel('Pollution Value')
        plt.ylabel('Count')
        plt.tight_layout()
        plt.show()

    elif choice == '5':

        plt.figure(figsize=(8, 8))

        data['Pollutant'].value_counts().head(5).plot(
            kind='pie',
            autopct='%1.1f%%',
            startangle=90)
        plt.title('Top 5 Pollutants')
        plt.ylabel('')
        plt.tight_layout()
        plt.show()

    elif choice == '6':
        numeric_data = data[['Value']].copy()
        numeric_data['Record Index'] = data.index

        plt.figure(figsize=(8, 6))
        sns.heatmap(numeric_data.corr(), annot=True, cmap='coolwarm')
        plt.title('Correlation Heatmap')
        plt.tight_layout()
        plt.show()

    elif choice == '7':

        plt.figure(figsize=(12, 8))

        plt.subplot(2, 2, 1)
        sns.histplot(data['Value'], bins=30)
        plt.title('Histogram')

        plt.subplot(2, 2, 2)
        data['Pollutant'].value_counts().head(5).plot(kind='bar', color='orange')
        plt.title('Top 5 Pollutants')

        plt.subplot(2, 2, 3)
        top10 = data.groupby('Country Label')['Value'].mean().sort_values(ascending=False).head(10)
        top10.plot(kind='bar', color='green')
        plt.title('Top 10 Countries')

        plt.subplot(2, 2, 4)
        sns.scatterplot(x=data.index, y='Value', data=data, alpha=0.3)
        plt.title('Scatter Plot')
        plt.tight_layout()
        plt.show()

    else:
        print("Wrong Choice")

except :
    print("Load Dataset First !")
    
```

```python
print("========== ANALYSIS RESULTS ==========")
print("1. Display Overall Summary")
print("2. Display Air Quality Performance")
print("3. Display Country Analysis")
print("4. Display Trend Analysis")
print("5. Display Final Report")


try:
    choice = input("Enter Your Choice: ")

    if choice == "1":
        print("========== OVERALL SUMMARY ==========")
        print("Total Records:", len(data))
        print("Total Countries:", data["Country"].nunique())
        print("Total Cities:", data["City"].nunique())
        print("Total Pollutants:", data["Pollutant"].nunique())
        print("Average Pollution Value:", data["Value"].mean())
        print("Highest Pollution Value:", data["Value"].max())
        print("Lowest Pollution Value:", data["Value"].min())

    elif choice == "2":
        print("========== AIR QUALITY PERFORMANCE ==========")
        highest_value_city = data.loc[data["Value"].idxmax(), "City"]
        lowest_value_city = data.loc[data["Value"].idxmin(), "City"]

        print("City with Highest Pollution Value:", highest_value_city)
        print("Highest Pollution Value:", data["Value"].max())

        print("City with Lowest Pollution Value:", lowest_value_city)
        print("Lowest Pollution Value:", data["Value"].min())

        print("Most Common Pollutant:", data["Pollutant"].value_counts().idxmax())

    elif choice == "3":
        print("========== COUNTRY ANALYSIS ==========")
        country_pollution = (data.groupby("Country")["Value"].mean().sort_values(ascending=False))

        print(country_pollution.head(10))
        print("Country with Highest Average Pollution:", country_pollution.idxmax())
        print("Country with Lowest Average Pollution:", country_pollution.idxmin())

    elif choice == "4":
        print("========== TREND ANALYSIS ==========")
        data["Last_Updated"] = pd.to_datetime(data["Last_Updated"], errors="coerce")

        daily_pollution = data.groupby("Last_Updated")["Value"].mean()
        print(daily_pollution)

        print("Date with Highest Average Pollution:", daily_pollution.idxmax())
        print("Date with Lowest Average Pollution:", daily_pollution.idxmin())

    elif choice == "5":
        print("========== FINAL AIR QUALITY REPORT ==========")
        print(data[["Value"]].describe())

        print("Overall Analysis:")

        if data["Value"].mean() > 0:
            print("Pollution values were recorded in this dataset.")
        else:
            print("No valid pollution values were recorded in this dataset.")

        if data["Pollutant"].nunique() > 1:
            print("Multiple pollutants were recorded in this dataset.")
        else:
            print("Only one pollutant was recorded in this dataset.")

        if data["Country"].nunique() > 1:
            print("Air quality data is available for multiple countries.")
        else:
            print("Air quality data is available for only one country.")

    else:
        print("Invalid Choice!")

except :
    print("Load Dataset First")
```
